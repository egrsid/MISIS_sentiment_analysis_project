```python
import numpy as np
import pandas as pd
import random
```

```python
def generate_positive_message():
    """
    Генерирует позитивное сообщение с разной степенью позитивности и длиной
    Возвращает строку с сообщением
    """
    
    # Уровни позитивности: 1 - слегка позитивное, 2 - умеренно позитивное, 3 - очень позитивное
    positivity_level = random.choices([1, 2, 3], weights=[0.3, 0.4, 0.3])[0]
    
    # Разная длина сообщений: короткие, средние, длинные
    length_type = random.choices(['short', 'medium', 'long'], weights=[0.3, 0.4, 0.3])[0]
    
    # Словари для разных уровней позитивности
    positive_adjectives_light = ["good", "nice", "fine", "decent", "satisfactory", "acceptable", "okay"]
    positive_adjectives_medium = ["great", "helpful", "pleasant", "efficient", "polite", "timely", "useful"]
    positive_adjectives_strong = ["excellent", "outstanding", "fantastic", "amazing", "exceptional", "superb", "wonderful", "brilliant"]
    
    positive_verbs_light = ["liked", "appreciated", "was pleased with"]
    positive_verbs_medium = ["appreciated", "valued", "was happy with", "enjoyed"]
    positive_verbs_strong = ["loved", "was thrilled with", "was delighted by", "was blown away by", "was extremely pleased with"]
    
    services = ["service", "support", "assistance", "help", "customer care", "response", "attention"]
    company_terms = ["your company", "the team", "your staff", "your representative", "this service", "your support"]
    issues = ["my issue", "the problem", "my request", "my inquiry", "the situation", "my case"]
    
    # Выбор слов в зависимости от уровня позитивности
    if positivity_level == 1:
        adjective = random.choice(positive_adjectives_light)
        verb = random.choice(positive_verbs_light)
        intensifiers = ["", "quite ", "rather "]
    elif positivity_level == 2:
        adjective = random.choice(positive_adjectives_medium)
        verb = random.choice(positive_verbs_medium)
        intensifiers = ["", "very ", "really "]
    else:  # level 3
        adjective = random.choice(positive_adjectives_strong)
        verb = random.choice(positive_verbs_strong)
        intensifiers = ["absolutely ", "extremely ", "incredibly ", "exceptionally "]
    
    intensifier = random.choice(intensifiers)
    
    # Короткие сообщения (5-15 слов)
    short_templates = [
        f"Thank you for the {adjective} {random.choice(services)}.",
        f"I {verb} the {random.choice(services)}.",
        f"{random.choice(['Good job!', 'Well done!', 'Nice work!'])}",
        f"The {random.choice(services)} was {intensifier}{adjective}.",
        f"I'm satisfied with the {random.choice(services)}.",
        f"Keep up the {adjective} work!",
        f"Thanks for your {adjective} help.",
        f"Quick and {adjective} response.",
        f"Problem solved {adjective}.",
        f"{random.choice(company_terms)} did a {adjective} job.",
    ]
    
    # Средние сообщения (10-25 слов)
    medium_templates = [
        f"I wanted to thank you for the {intensifier}{adjective} {random.choice(services)} I received. {random.choice(['It really helped.', 'Much appreciated.', 'Thank you!'])}",
        f"Your team provided {intensifier}{adjective} {random.choice(services)} regarding {random.choice(issues)}. {random.choice(['I am pleased.', 'Very professional.', 'Good experience.'])}",
        f"I {verb} how you handled {random.choice(issues)}. The {random.choice(services)} was {intensifier}{adjective} and efficient.",
        f"Thanks to {random.choice(company_terms)} for the {adjective} assistance. {random.choice(['Will recommend.', 'Very impressed.', 'Great work!'])}",
        f"The {random.choice(services)} I got from {random.choice(company_terms)} was {intensifier}{adjective}. {random.choice(['Good communication.', 'Timely response.', 'Helpful advice.'])}",
        f"I'm writing to express my satisfaction with the {random.choice(services)}. {random.choice(company_terms)} was {intensifier}{adjective}.",
        f"My experience with {random.choice(company_terms)} was {intensifier}{adjective}. {random.choice(['Thanks again.', 'Highly satisfied.', 'Great support.'])}",
    ]
    
    # Длинные сообщения (20-50 слов)
    long_templates = [
        f"I wanted to take a moment to express my appreciation for the {intensifier}{adjective} {random.choice(services)} I recently received from {random.choice(company_terms)}. The attention to detail and professional approach made a significant difference in resolving {random.choice(issues)}. {random.choice(['I will certainly recommend your services to others.', 'This level of service is commendable.', 'Thank you for going above and beyond.'])}",
        f"After experiencing the {random.choice(services)} from {random.choice(company_terms)}, I must say I was {intensifier}{adjective} impressed. The way you handled {random.choice(issues)} demonstrated real expertise and care. {random.choice(['It\'s refreshing to see such dedication.', 'Your team\'s efforts did not go unnoticed.', 'I feel valued as a customer.'])}",
        f"In today's world where good {random.choice(services)} is hard to find, I was pleasantly surprised by the {intensifier}{adjective} experience I had with {random.choice(company_terms)}. From start to finish, the process was smooth and the communication was clear. {random.choice(['This sets a high standard for others.', 'My expectations were exceeded.', 'A truly professional operation.'])}",
        f"I've been reflecting on the {random.choice(services)} I received regarding {random.choice(issues)}, and I feel compelled to share how {intensifier}{adjective} it was. {random.choice(company_terms)} showed genuine concern and worked diligently to ensure a positive outcome. {random.choice(['This kind of service builds lasting loyalty.', 'You have earned my trust and future business.', 'Exceptional work from start to finish.'])}",
    ]
    
    # Выбор шаблона в зависимости от желаемой длины
    if length_type == 'short':
        template = random.choice(short_templates)
    elif length_type == 'medium':
        template = random.choice(medium_templates)
    else:  # long
        template = random.choice(long_templates)
    
    # Добавление вариативности с вероятностью 30%
    if random.random() < 0.3:
        additions = [
            f" {random.choice(['Thank you!', 'Much appreciated!', 'Greatly appreciated!'])}",
            f" {random.choice(['I\'ll be back.', 'Will use again.', 'Definitely returning.'])}",
            f" {random.choice(['Five stars!', 'Top marks!', 'Highest rating!'])}",
            f" {random.choice(['Better than expected.', 'Exceeded expectations.', 'Went the extra mile.'])}",
        ]
        template += random.choice(additions)
    
    return template
```

```python
def generate_negative_message():
    """
    Генерирует негативное сообщение с разной степенью негативности и длиной
    Возвращает строку с сообщением
    """
    
    # Уровни негативности: 1 - слегка негативное, 2 - умеренно негативное, 3 - очень негативное
    negativity_level = random.choices([1, 2, 3], weights=[0.3, 0.4, 0.3])[0]
    
    # Разная длина сообщений: короткие, средние, длинные
    length_type = random.choices(['short', 'medium', 'long'], weights=[0.3, 0.4, 0.3])[0]
    
    # Словари для разных уровней негативности
    negative_adjectives_light = ["poor", "unsatisfactory", "subpar", "disappointing", "below average", "mediocre"]
    negative_adjectives_medium = ["bad", "unacceptable", "frustrating", "inadequate", "unprofessional", "slow"]
    negative_adjectives_strong = ["terrible", "awful", "horrible", "atrocious", "dreadful", "appalling", "disgusting"]
    
    negative_verbs_light = ["was not happy with", "was disappointed by", "found issues with"]
    negative_verbs_medium = ["was frustrated by", "was upset with", "had problems with", "was dissatisfied with"]
    negative_verbs_strong = ["was outraged by", "was furious about", "was appalled by", "was shocked by", "was disgusted with"]
    
    services = ["service", "support", "assistance", "help", "customer care", "response", "attention"]
    company_terms = ["your company", "the team", "your staff", "your representative", "this service", "your support"]
    issues = ["my issue", "the problem", "my request", "my inquiry", "the situation", "my case"]
    
    # Выбор слов в зависимости от уровня негативности
    if negativity_level == 1:
        adjective = random.choice(negative_adjectives_light)
        verb = random.choice(negative_verbs_light)
        intensifiers = ["", "rather ", "somewhat "]
    elif negativity_level == 2:
        adjective = random.choice(negative_adjectives_medium)
        verb = random.choice(negative_verbs_medium)
        intensifiers = ["", "very ", "really "]
    else:  # level 3
        adjective = random.choice(negative_adjectives_strong)
        verb = random.choice(negative_verbs_strong)
        intensifiers = ["absolutely ", "extremely ", "incredibly ", "completely "]
    
    intensifier = random.choice(intensifiers)
    
    # Базовые шаблоны
    templates = []
    
    # Короткие сообщения (5-15 слов)
    short_templates = [
        f"I'm not happy with the {adjective} {random.choice(services)}.",
        f"The {random.choice(services)} was {intensifier}{adjective}.",
        f"{random.choice(['Very disappointed.', 'Not good enough.', 'Needs improvement.'])}",
        f"Poor {random.choice(services)} from {random.choice(company_terms)}.",
        f"I {verb} the {random.choice(services)}.",
        f"Unsatisfactory experience.",
        f"Not what I expected.",
        f"{random.choice(company_terms)} needs to do better.",
        f"Bad experience with {random.choice(services)}.",
        f"Dissatisfied customer here.",
    ]
    
    # Средние сообщения (10-25 слов)
    medium_templates = [
        f"I wanted to express my disappointment with the {intensifier}{adjective} {random.choice(services)} I received. {random.choice(['Not acceptable.', 'Very poor experience.', 'Will not recommend.'])}",
        f"Your team provided {intensifier}{adjective} {random.choice(services)} regarding {random.choice(issues)}. {random.choice(['I am not satisfied.', 'Very unprofessional.', 'Bad experience.'])}",
        f"I {verb} how you handled {random.choice(issues)}. The {random.choice(services)} was {intensifier}{adjective} and inefficient.",
        f"Unfortunately, my experience with {random.choice(company_terms)} was {intensifier}{adjective}. {random.choice(['Very disappointed.', 'Not good enough.', 'Needs serious improvement.'])}",
        f"The {random.choice(services)} I got from {random.choice(company_terms)} was {intensifier}{adjective}. {random.choice(['Poor communication.', 'Slow response.', 'Unhelpful advice.'])}",
        f"I'm writing to express my dissatisfaction with the {random.choice(services)}. {random.choice(company_terms)} was {intensifier}{adjective}.",
        f"My experience with {random.choice(company_terms)} was {intensifier}{adjective}. {random.choice(['Very frustrated.', 'Highly dissatisfied.', 'Poor support.'])}",
    ]
    
    # Длинные сообщения (20-50 слов)
    long_templates = [
        f"I wanted to take a moment to express my frustration with the {intensifier}{adjective} {random.choice(services)} I recently received from {random.choice(company_terms)}. The lack of attention to detail and unprofessional approach created significant problems in resolving {random.choice(issues)}. {random.choice(['I will certainly warn others about your services.', 'This level of service is unacceptable.', 'I expect better from a professional company.'])}",
        f"After experiencing the {random.choice(services)} from {random.choice(company_terms)}, I must say I was {intensifier}{adjective} disappointed. The way you handled {random.choice(issues)} demonstrated a real lack of expertise and care. {random.choice(['It\'s disappointing to see such poor service.', 'Your team\'s performance was unacceptable.', 'I feel completely disregarded as a customer.'])}",
        f"In today's competitive market, the {intensifier}{adjective} experience I had with {random.choice(company_terms)} stands out for all the wrong reasons. From start to finish, the process was frustrating and the communication was unclear. {random.choice(['This falls far below industry standards.', 'My expectations were not met in any way.', 'A truly disappointing operation.'])}",
        f"I've been reflecting on the {random.choice(services)} I received regarding {random.choice(issues)}, and I feel compelled to share how {intensifier}{adjective} it was. {random.choice(company_terms)} showed no genuine concern and failed to work diligently toward a resolution. {random.choice(['This kind of service destroys customer loyalty.', 'You have lost my trust and future business.', 'Poor work from start to finish.'])}",
    ]
    
    # Выбор шаблона в зависимости от желаемой длины
    if length_type == 'short':
        template = random.choice(short_templates)
    elif length_type == 'medium':
        template = random.choice(medium_templates)
    else:  # long
        template = random.choice(long_templates)
    
    # Добавление вариативности с вероятностью 30%
    if random.random() < 0.3:
        additions = [
            f" {random.choice(['Never again.', 'Will not return.', 'Taking my business elsewhere.'])}",
            f" {random.choice(['Zero stars!', 'Worst experience!', 'Bottom rating!'])}",
            f" {random.choice(['Worse than expected.', 'Failed to meet basic standards.', 'Complete letdown.'])}",
            f" {random.choice(['Formal complaint to follow.', 'Demanding a refund.', 'Expecting compensation.'])}",
        ]
        template += random.choice(additions)
    
    return template
```

```python
# Генерация 1000 примеров (500 негативных, 300 позитивных, 200 нейтральных)
np.random.seed(42)  # Для воспроизводимости
random.seed(42)

n_samples = 1000
n_negative = 500
n_positive = 500

messages = []
labels = []
```

```python
# Генерация негативных сообщений
for _ in range(n_negative):
    messages.append(generate_negative_message())
    labels.append(0)  # 0 = негативное

# Генерация позитивных сообщений
for _ in range(n_positive):
    messages.append(generate_positive_message())
    labels.append(1)  # 1 = позитивное
```

```python
# Перемешиваем данные
indices = np.random.permutation(len(messages))
messages = np.array(messages)[indices]
labels = np.array(labels)[indices]

# Создаем DataFrame
df = pd.DataFrame({
    'message': messages,
    'sentiment': labels
})
```

```python
# Добавляем метки для читаемости
sentiment_map = {0: 'negative', 1: 'positive'}
df['sentiment_label'] = df['sentiment'].map(sentiment_map)

# Сохраняем в CSV
csv_filename = f'DataBases/external_testing_dataset.csv'
df.to_csv(csv_filename, index=False, encoding='utf-8')

# Выводим статистику
print(f"Датасет успешно создан и сохранен в файл: {csv_filename}")
print("\nСтатистика датасета:")
print(f"Всего сообщений: {len(df)}")
print(f"Негативных (0): {sum(df['sentiment'] == 0)}")
print(f"Позитивных (1): {sum(df['sentiment'] == 1)}")
print("\nПервые 10 примеров:")
print(df[['message', 'sentiment_label']].head(10).to_string(index=False))
```
